```python

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Data evaluasi klasifikasi
classification_data = {
    "Logistic Regression": {
        "Accuracy": 0.9934,
        "Precision Macro": 0.9919,
        "Recall Macro": 0.9908,
        "F1 Macro": 0.9913
    },
    "Decision Tree Classifier": {
        "Accuracy": 0.9765,
        "Precision Macro": 0.9676,
        "Recall Macro": 0.9709,
        "F1 Macro": 0.9692
    },
    "KNN (k=5)": {
        "Accuracy": 0.9934,
        "Precision Macro": np.nan,
        "Recall Macro": np.nan,
        "F1 Macro": np.nan
    },
    "KNN (k=1, Tuned)": {
        "Accuracy": 0.9953,
        "Precision Macro": 0.9944,
        "Recall Macro": 0.9932,
        "F1 Macro": 0.9938
    }
}

df_classification = pd.DataFrame(
    classification_data
).T

# Warna berbeda untuk setiap model
colors = {
    "Logistic Regression": "blue",
    "Decision Tree Classifier": "red",
    "KNN (k=5)": "green",
    "KNN (k=1, Tuned)": "purple"
}

plt.figure(figsize=(12, 6))

for model in df_classification.index:
    plt.plot(
        df_classification.columns,
        df_classification.loc[model],
        marker="o",
        linewidth=2,
        markersize=7,
        color=colors[model],
        label=model
    )

plt.title(
    "Perbandingan Kinerja Model Klasifikasi Wine",
    fontsize=14
)
plt.xlabel("Evaluation Metrics")
plt.ylabel("Score")
plt.ylim(0.95, 1.00)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(title="Model")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python

# Data evaluasi regresi dari data uji
regression_data = {
    "Linear Regression": {
        "MAE": 0.5673,
        "RMSE": 0.7335,
        "R2": 0.3053
    },
    "Decision Tree Regressor": {
        "MAE": 0.6739,
        "RMSE": 0.9834,
        "R2": -0.2486
    },
    "Random Forest": {
        "MAE": 0.5336,
        "RMSE": 0.6908,
        "R2": 0.3838
    },
    "Random Forest Tuned": {
        "MAE": 0.5265,
        "RMSE": 0.6807,
        "R2": 0.4019
    }
}

df_regression = pd.DataFrame(
    regression_data
).T

# Normalisasi setiap metrik agar dapat divisualisasikan bersama
df_regression_score = df_regression.copy()

for metric in ["MAE", "RMSE"]:
    min_value = df_regression[metric].min()
    max_value = df_regression[metric].max()

    df_regression_score[metric] = (
        (max_value - df_regression[metric])
        / (max_value - min_value)
    )

# R2: semakin tinggi, semakin baik
min_r2 = df_regression["R2"].min()
max_r2 = df_regression["R2"].max()

df_regression_score["R2"] = (
    (df_regression["R2"] - min_r2)
    / (max_r2 - min_r2)
)

# Warna berbeda untuk setiap model
regression_colors = {
    "Linear Regression": "blue",
    "Decision Tree Regressor": "red",
    "Random Forest": "green",
    "Random Forest Tuned": "purple"
}

plt.figure(figsize=(12, 6))

for model in df_regression_score.index:
    plt.plot(
        df_regression_score.columns,
        df_regression_score.loc[model],
        marker="o",
        linewidth=2,
        markersize=7,
        color=regression_colors[model],
        label=model
    )

plt.title(
    "Perbandingan Model Regresi Kualitas Wine",
    fontsize=14
)
plt.xlabel("Evaluation Metrics")
plt.ylabel("Normalized Relative Score (0–1)")
plt.ylim(-0.05, 1.05)
plt.grid(True, linestyle="--", alpha=0.4)
plt.legend(title="Model")
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np

metrics = [
    "Accuracy",
    "Precision Macro",
    "Recall Macro",
    "F1 Macro"
]

models = [
    "Logistic Regression",
    "Decision Tree Classifier",
    "KNN (k=5)",
    "KNN (k=1, Tuned)"
]

classification_scores = {
    "Logistic Regression": [0.9934, 0.9919, 0.9908, 0.9913],
    "Decision Tree Classifier": [0.9765, 0.9676, 0.9709, 0.9692],
    "KNN (k=5)": [0.9934, np.nan, np.nan, np.nan],
    "KNN (k=1, Tuned)": [0.9953, 0.9944, 0.9932, 0.9938]
}

colors = {
    "Logistic Regression": "royalblue",
    "Decision Tree Classifier": "tomato",
    "KNN (k=5)": "seagreen",
    "KNN (k=1, Tuned)": "mediumpurple"
}

fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes = axes.flatten()

for i, metric in enumerate(metrics):
    ax = axes[i]

    for j, model in enumerate(models):
        score = classification_scores[model][i]

        if not np.isnan(score):
            ax.barh(
                j,
                score * 100,
                color=colors[model],
                height=0.65
            )
            ax.text(
                score * 100 + 0.03,
                j,
                f"{score * 100:.2f}%",
                va="center",
                fontsize=9
            )

    ax.set_title(metric)
    ax.set_yticks(range(len(models)))
    ax.set_yticklabels(models, fontsize=8)
    ax.invert_yaxis()
    ax.set_xlim(95, 100.5)
    ax.set_xlabel("Score (%)")
    ax.grid(axis="x", linestyle="--", alpha=0.3)
    ax.set_axisbelow(True)

fig.suptitle(
    "Perbandingan Model Klasifikasi Jenis Wine",
    fontsize=15
)
plt.tight_layout()
plt.show()
```

```python

regression_scores = {
    "Linear Regression": {
        "MAE": 0.5673,
        "RMSE": 0.7335,
        "R2": 0.3053
    },
    "Decision Tree Regressor": {
        "MAE": 0.6739,
        "RMSE": 0.9834,
        "R2": -0.2486
    },
    "Random Forest": {
        "MAE": 0.5336,
        "RMSE": 0.6908,
        "R2": 0.3838
    },
    "Random Forest Tuned": {
        "MAE": 0.5265,
        "RMSE": 0.6807,
        "R2": 0.4019
    }
}

regression_colors = {
    "Linear Regression": "royalblue",
    "Decision Tree Regressor": "tomato",
    "Random Forest": "seagreen",
    "Random Forest Tuned": "mediumpurple"
}

regression_metrics = ["MAE", "RMSE", "R2"]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, metric in enumerate(regression_metrics):
    ax = axes[i]

    names = list(regression_scores.keys())
    values = [
        regression_scores[name][metric]
        for name in names
    ]
    bar_colors = [
        regression_colors[name]
        for name in names
    ]

    bars = ax.barh(
        names,
        values,
        color=bar_colors,
        height=0.65
    )

    ax.axvline(0, color="black", linewidth=0.8)
    ax.set_title(metric)
    ax.set_xlabel("Nilai Evaluasi")
    ax.invert_yaxis()
    ax.grid(axis="x", linestyle="--", alpha=0.3)
    ax.set_axisbelow(True)

    for bar, value in zip(bars, values):
        offset = 0.015 if value >= 0 else -0.015
        ax.text(
            value + offset,
            bar.get_y() + bar.get_height() / 2,
            f"{value:.4f}",
            va="center",
            ha="left" if value >= 0 else "right",
            fontsize=8
        )

fig.suptitle(
    "Perbandingan Model Regresi Kualitas Wine",
    fontsize=15
)
plt.tight_layout()
plt.show()
```
